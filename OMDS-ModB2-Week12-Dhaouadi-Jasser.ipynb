# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

**Answer**

**Separate plots (side-by-side)**

- **Advantages**:

    - Easier to read each week individually
    - Less clutter => clearer trends within each week
    - Good for focusing on pattern within a single week

- **Disadvantages**:

    - Harder to compare the two weeks directly
    - You have to visually switch between plots



**Same plot (overlaid lines)**

- **Advantages**:

    - Easy to compare the two weeks directly
    - Differences and similarities are immediately visible
    - Better for spotting changes across weeks

- **Disadvantages**:

    - Can get cluttered if many lines
    - Lines may overlap, making it harder to read individual values

The graph selected for reproduction is **Figure 8.19 from page 203 of the book (Introduction of Storytelling With Data)**.

```python
import matplotlib.pyplot as plt
import numpy as np

# Years
years = np.array([2008,2009,2010,2011,2012,2013,2014])

# Approximate data
A = [395, 420, 425, 390, 300, 270, 260]
B = [360, 400, 410, 380, 290, 260, 250]
C = [None, None, 120, 180, 200, 240, 210]
D = [None, None, None, 170, 260, 220, 215]
E = [None, None, None, None, None, 120, 220]

avg_point = (2014, 223)

fig, ax = plt.subplots(figsize=(10,6))

# --- Recommended band ---
ax.axhspan(150, 200, color="#D3D3D3", alpha=0.8)
ax.text(2007.1, 175, "Recommended range", fontsize=18, color="#244a75")

# --- Lines ---
for line in [A, B, C, D, E]:
    ax.plot(years, line, color="#9e9e9e", linewidth=5)

# --- Labels ---
ax.text(2007.6, 390, "A", fontsize=20, color="dimgray")
ax.text(2007.6, 345, "B", fontsize=20, color="dimgray")
ax.text(2009.6, 110, "C", fontsize=20, color="dimgray")
ax.text(2010.7, 140, "D", fontsize=20, color="dimgray")
ax.text(2012.7, 105, "E", fontsize=20, color="dimgray")

# --- Average point ---
ax.scatter(*avg_point, color="#244a75", s=200, zorder=5)
ax.text(2013.6, 225, "AVG", color="#244a75", fontsize=16)

# --- Left-aligned headline ---
fig.text(
    0.01, 1.15,
    "To be competitive, we recommend introducing our product below\n"
    "the $223 average price point in the $150–$200 range",
    fontsize=20, color="#244a75", ha='left', va='top'
)

# --- Subtitle ---
ax.set_title("Retail price over time", fontsize=18, loc='left')

# --- Axes ---
ax.set_xlim(2007, 2014)
ax.set_ylim(0, 500)

ax.set_xlabel("Year", fontsize=14)
ax.set_ylabel("Average price", fontsize=14)

yticks = [0,150,200,300,400,500]
ax.set_yticks(yticks)
ax.set_yticklabels([f"${y}" for y in yticks])

# ✅ Highlight 150 and 200
for label, value in zip(ax.get_yticklabels(), yticks):
    if value in [150, 200]:
        label.set_color("#244a75")
        label.set_fontweight("bold")

# --- Style ---
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```
