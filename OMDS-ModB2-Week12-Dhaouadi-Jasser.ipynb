# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

**Answer**

**Separate plots (side-by-side)**

- **Advantages**:

    - Easier to read each week individually
    - Less clutter => clearer trends within each week
    - Good for focusing on pattern within a single week

- **Disadvantages**:

    - Harder to compare the two weeks directly
    - You have to visually switch between plots



**Same plot (overlaid lines)**

- **Advantages**:

    - Easy to compare the two weeks directly
    - Differences and similarities are immediately visible
    - Better for spotting changes across weeks

- **Disadvantages**:

    - Can get cluttered if many lines
    - Lines may overlap, making it harder to read individual values

```python
# Imported Packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.manifold import TSNE
```

## Dataset 1: CDC Diabetes Health Indicators  (sourced from BRFSS) 

#### [Dataset 1] | *1- Load Preprocessed Dataset*

```python
dataset_path = "data/dataset1_diabetes_preprocessed_week2.csv"
df_diabetes = pd.read_csv(dataset_path)

# drop index-like column
df_diabetes = df_diabetes.drop(columns=["Unnamed: 0"])
```

#### [Dataset 1] | *2- PCA*

```python
target_col = "Diabetes_012"

X = df_diabetes.drop(columns=[target_col])
y = df_diabetes[target_col]

feature_names = X.columns.tolist()

# -----------------------------
# 2. Standardize predictors
# -----------------------------
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# -----------------------------
# 3. Fit PCA
# -----------------------------
pca = PCA()
X_pca = pca.fit_transform(X_scaled)

# -----------------------------
# 4. Principal component vectors
# -----------------------------
pc_names = [f"PC{i+1}" for i in range(len(feature_names))]
df_components = pd.DataFrame(
    pca.components_,
    index=pc_names,
    columns=feature_names
)

print("Principal Component Vectors (Loadings):")
display(df_components)

# -----------------------------
# 5. Explained variance
# -----------------------------
explained_var = pd.DataFrame({
    "Principal Component": pc_names,
    "Explained Variance Ratio": pca.explained_variance_ratio_,
    "Cumulative Variance": np.cumsum(pca.explained_variance_ratio_)
})

print("Explained Variance by Component:")
display(explained_var)

# Scree plot
plt.figure(figsize=(8,5))
plt.plot(range(1, len(feature_names)+1), pca.explained_variance_ratio_, marker='o')
plt.xlabel("Principal Component")
plt.ylabel("Explained Variance Ratio")
plt.title("Scree Plot for Dataset 1 (Diabetes) PCA")
plt.show()

# Cumulative variance plot
plt.figure(figsize=(8,5))
plt.plot(range(1, len(feature_names)+1), np.cumsum(pca.explained_variance_ratio_), marker='o')
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Cumulative Explained Variance (Diabetes Dataset)")
plt.ylim(0, 1.05)
plt.show()

# -----------------------------
# 6. PCA scatter plot (PC1 vs PC2)
# -----------------------------
pca_df = pd.DataFrame({
    "PC1": X_pca[:, 0],
    "PC2": X_pca[:, 1],
    "Diabetes_012": y
})

plt.figure(figsize=(8,6))
sns.scatterplot(
    data=pca_df.sample(min(8000, len(pca_df)), random_state=0),  # sample for speed/clarity
    x="PC1",
    y="PC2",
    hue="Diabetes_012",
    alpha=0.6,
    s=30
)
plt.title("PCA Scatter Plot: PC1 vs PC2 (Diabetes Dataset)")
plt.show()

# -----------------------------
# 7. Biplot (PC1 and PC2)
# -----------------------------
# scores for plotting
scores = X_pca[:, :2]

# loadings for arrows
loadings = pca.components_.T[:, :2]

# scale arrows for visibility
arrow_scale = 3.0

plt.figure(figsize=(10,8))
plt.scatter(
    scores[:8000, 0], scores[:8000, 1],  # subset for readability
    c=y.iloc[:8000],
    cmap="viridis",
    alpha=0.35,
    s=18
)

for i, feature in enumerate(feature_names):
    plt.arrow(
        0, 0,
        loadings[i, 0] * arrow_scale,
        loadings[i, 1] * arrow_scale,
        color="red",
        alpha=0.8,
        head_width=0.08
    )
    plt.text(
        loadings[i, 0] * arrow_scale * 1.12,
        loadings[i, 1] * arrow_scale * 1.12,
        feature,
        color="darkred",
        fontsize=9
    )

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA Biplot for Diabetes Dataset")
plt.axhline(0, color="gray", linewidth=0.8)
plt.axvline(0, color="gray", linewidth=0.8)
plt.show()

# -----------------------------
# 8. Top contributing variables for PC1 and PC2
# -----------------------------
pc1_top = df_components.loc["PC1"].abs().sort_values(ascending=False).head(10)
pc2_top = df_components.loc["PC2"].abs().sort_values(ascending=False).head(10)

print("Top variables contributing to PC1:")
display(pc1_top)

print("Top variables contributing to PC2:")
display(pc2_top)
```

#### [Dataset 1] | *3- PCA Interpretation*


**3.1 Explained Variance**

- **PC1 explains ~17%** of the variance, and PC2 adds ~8%  
- Variance is **spread across many components** (no sharp drop after first few PCs)

=> The dataset is **not low-dimensional** → multiple factors drive diabetes risk

<br>

**3.2 PCA Scatter Plot**

- The three classes **overlap heavily**
- No clear clusters or separation in PC1–PC2 space  

=> Diabetes groups are **not linearly separable**

<br>

**3.3 Principal Components Meaning**

**PC1 (main axis):**
- Driven by:
  - **GenHlth, DiffWalk, PhysHlth, Income, Education**
- Represents a **general health & socioeconomic condition axis**

**PC2 (secondary axis):**
- Driven by:
  - **Age, Healthcare access, Cost barriers, HighBP/Chol**
- Represents an **age & healthcare access axis**

<br>

**3.4 Biplot Insights**

- Health-related variables (GenHlth, PhysHlth, DiffWalk) point in similar directions

**3.6 Final Takeaway**

## Dataset 2: Estimation of Obesity Levels Based on Eating Habits and Physical Condition

#### [Dataset 2] | *1- Load Preprocessed Dataset*

```python
df_obesity = pd.read_csv("data/dataset2_obesity_preprocessed_week2_1.csv")

df_obesity = df_obesity.drop(columns=["Unnamed: 0"])
```

#### [Dataset 2] | *2- PCA*

```python
# -----------------------------
# 2. UPDATED TARGET MAPPING
# -----------------------------
def map_obesity(x):
    if x in ["Insufficient_Weight", "Normal_Weight", "Overweight_Level_I", "Overweight_Level_II"]:
        return "Not Obese"
    elif x == "Obesity_Type_I":
        return "Obesity Type I"
    elif x == "Obesity_Type_II":
        return "Obesity Type II"
    else:
        return "Obesity Type III"

df_obesity["Obesity_Group"] = df_obesity["NObeyesdad"].apply(map_obesity)

X = df_obesity.drop(columns=["NObeyesdad", "Obesity_Group"])
y = df_obesity["Obesity_Group"]

# -----------------------------
# 3. One-hot encoding
# -----------------------------
X_encoded = pd.get_dummies(X, drop_first=True)

feature_names = X_encoded.columns.tolist()

# -----------------------------
# 4. Standardize
# -----------------------------
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_encoded)

# -----------------------------
# 5. PCA
# -----------------------------
pca = PCA()
X_pca = pca.fit_transform(X_scaled)

# -----------------------------
# 6. PCA scatter (UPDATED TARGET)
# -----------------------------
pca_df = pd.DataFrame({
    "PC1": X_pca[:, 0],
    "PC2": X_pca[:, 1],
    "Obesity_Group": y
})

plt.figure(figsize=(8,6))
sns.scatterplot(
    data=pca_df.sample(min(4000, len(pca_df)), random_state=0),
    x="PC1",
    y="PC2",
    hue="Obesity_Group",
    alpha=0.7,
    s=40
)

plt.title("PCA Scatter (Dataset 2: Obesity Levels - 4 Classes)")
plt.show()

# -----------------------------
# 7. Biplot (same as before)
# -----------------------------
scores = X_pca[:, :2]
loadings = pca.components_.T[:, :2]

arrow_scale = 3

plt.figure(figsize=(10,8))
plt.scatter(
    scores[:4000, 0],
    scores[:4000, 1],
    c=pd.factorize(y[:4000])[0],
    cmap="viridis",
    alpha=0.4,
    s=20
)

for i, feature in enumerate(feature_names[:20]):
    plt.arrow(
        0, 0,
        loadings[i, 0]*arrow_scale,
        loadings[i, 1]*arrow_scale,
        color="red",
        alpha=0.8
    )
    plt.text(
        loadings[i, 0]*arrow_scale*1.1,
        loadings[i, 1]*arrow_scale*1.1,
        feature,
        fontsize=8
    )

plt.axhline(0, color="gray")
plt.axvline(0, color="gray")
plt.title("PCA Biplot (Dataset 2: Obesity Levels)")
plt.show()
```

#### [Dataset 2] | *3- PCA Interpretation*



**3.1 Overall Structure**
- PCA shows **clear separation between classes**, especially along **PC1**
- There is a clear progression:
  
=> **Not Obese → Type I → Type II → Type III**

<br>

**3.2 Key Insight**

- **PC1 captures obesity severity**
- Driven mainly by:
  - **BMI**
  - **Weight**

=> This explains the strong class separation

<br>

**3.3 PC2 Meaning**

- Captures **lifestyle variation** (diet, activity)
- Less important for class separation

<br>

**3.4 Important Note**

> Separation is very clean because BMI strongly determines obesity class

=> This makes the dataset:
- **Easy to model**
- Likely to produce **very high accuracy**

<br>

**3.5 Final Takeaway**

- PCA reveals a **strong, structured dataset**
- Classes are **naturally separable**
- Main driver = **BMI / body composition**

> Unlike Dataset 1 (diabetes), this problem (obesity) is **much simpler and more linear**

```python
import matplotlib.pyplot as plt
import numpy as np

# Years
years = np.array([2008,2009,2010,2011,2012,2013,2014])

# Approximate data
A = [395, 420, 425, 390, 300, 270, 260]
B = [360, 400, 410, 380, 290, 260, 250]
C = [None, None, 120, 180, 200, 240, 210]
D = [None, None, None, 170, 260, 220, 215]
E = [None, None, None, None, None, 120, 220]

avg_point = (2014, 223)

fig, ax = plt.subplots(figsize=(10,6))

# --- Recommended band ---
ax.axhspan(150, 200, color="#D3D3D3", alpha=0.8)
ax.text(2007.1, 175, "Recommended range", fontsize=18, color="#244a75")

# --- Lines ---
for line in [A, B, C, D, E]:
    ax.plot(years, line, color="#9e9e9e", linewidth=5)

# --- Labels ---
ax.text(2007.6, 390, "A", fontsize=20, color="dimgray")
ax.text(2007.6, 345, "B", fontsize=20, color="dimgray")
ax.text(2009.6, 110, "C", fontsize=20, color="dimgray")
ax.text(2010.7, 140, "D", fontsize=20, color="dimgray")
ax.text(2012.7, 105, "E", fontsize=20, color="dimgray")

# --- Average point ---
ax.scatter(*avg_point, color="#244a75", s=200, zorder=5)
ax.text(2013.6, 225, "AVG", color="#244a75", fontsize=16)

# --- Left-aligned headline ---
fig.text(
    0.01, 1.15,
    "To be competitive, we recommend introducing our product below\n"
    "the $223 average price point in the $150–$200 range",
    fontsize=20, color="#244a75", ha='left', va='top'
)

# --- Subtitle ---
ax.set_title("Retail price over time", fontsize=18, loc='left')

# --- Axes ---
ax.set_xlim(2007, 2014)
ax.set_ylim(0, 500)

ax.set_xlabel("Year", fontsize=14)
ax.set_ylabel("Average price", fontsize=14)

yticks = [0,150,200,300,400,500]
ax.set_yticks(yticks)
ax.set_yticklabels([f"${y}" for y in yticks])

# ✅ Highlight 150 and 200
for label, value in zip(ax.get_yticklabels(), yticks):
    if value in [150, 200]:
        label.set_color("#244a75")
        label.set_fontweight("bold")

# --- Style ---
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```
