# 2. Weekly graph question

Below are the graphs from last week's weekly graph question.  Suppose you wanted to cause the viewer to examine the last three months of data first.  How could you use preattentive attributes to do that?  Suggest several possibilities.

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size = numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='ME')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Nitrate level (% of eutrophic value)")
```

```python
np.random.seed(1)
time_series = 100 * (1 - (np.random.normal(size = numdates) / 8 + np.arange(numdates) / 16))
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='ME')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Phosphate level (% of eutrophic value)")
```

### Answer:

To make viewers look at the last three months first, we can use **preattentive attributes** such as <u>color, size, or contrast</u> so **those points stand out immediately.**

**1. Highlight the Last Three Months with Color**
- Use a neutral color for earlier data and a bright color for the last three months.

```python
plt.figure()
sns.lineplot(x=datearray, y=time_series, color='lightgray')
sns.lineplot(x=datearray[-3:], y=time_series[-3:], color='red', linewidth=3)
plt.xlabel("Year and month")
plt.ylabel("Nitrate level (% of eutrophic value)")
plt.show()
```

**2. Increase Line Thickness**

- Make the last three months thicker so they visually pop.

```python
plt.figure()
sns.lineplot(x=datearray, y=time_series, color='gray')
sns.lineplot(x=datearray[-3:], y=time_series[-3:], color='gray', linewidth=7)
plt.xlabel("Year and month")
plt.ylabel("Nitrate level (% of eutrophic value)")
```

**3. Add Large Markers**

- Add larger markers to the last three data points.

```python
plt.figure()
sns.lineplot(x=datearray, y=time_series)
plt.axvspan(datearray[-3], datearray[-1], color='yellow', alpha=0.3)
plt.xlabel("Year and month")
plt.ylabel("Nitrate level (% of eutrophic value)")
```

## Dataset 1: CDC Diabetes Health Indicators  (sourced from BRFSS) 

#### [Dataset 1] | *1- Load Preprocessed Dataset*

```python
dataset_path = "data/dataset1_diabetes_preprocessed_week2.csv"
df_diabetes = pd.read_csv(dataset_path)
```

#### [Dataset 1] | *2- General Health vs Diabetes Status*

```python
# Dataset 1: General Health vs Diabetes Status

plt.figure(figsize=(6,4))

sns.boxplot(
    data=df_diabetes,
    x="Diabetes_012",
    y="GenHlth"
)

plt.xticks([0,1,2], ["No Diabetes","Prediabetes","Diabetes"])
plt.xlabel("Diabetes Status")
plt.ylabel("General Health Score")
plt.title("General Health Status Across Diabetes Categories")

plt.tight_layout()
plt.show()
```

**Interpretation**

- Individuals with diabetes tend to report **poorer general health** compared to those without diabetes.

- The median health score increases (worse health) from **no diabetes to prediabetes to diabetes.**

- This suggests that diabetes is associated with **broader declines in perceived health and well-being**, not just metabolic indicators.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 99-119 of the Storytelling With Data book as best you can. (The first half of chapter four). You do not have to get the exact data values right, just the overall look and feel.

### Answer:
[**Note** : I used dummy data to reproduce the graph]
</br> </br>
The graph selected for reproduction is **Figure 4.8 from page 111 of the book (Introduction of Storytelling With Data)**.

```python
import matplotlib.pyplot as plt

# Data from the figure
concerns = [
    "Engine power is less than expected",
    "Tires make excessive noise while driving",
    "Engine makes abnormal/excessive noise",
    "Seat material concerns",
    "Excessive wind noise",
    "Hesitation or delay when shifting",
    "Bluetooth system has poor sound quality",
    "Steering system/wheel has too much play",
    "Bluetooth system is difficult to use",
    "Front seat audio/entertainment/navigation controls"
]

values = [12.9, 12.3, 11.6, 11.6, 11.0, 10.3, 10.0, 8.8, 8.6, 8.2]

# Colors (top 7 highlighted, bottom 3 gray)
colors = ["#C0504D"]*7 + ["#808080"]*3

# Reverse lists so the first item appears at the top
concerns = concerns[::-1]
values = values[::-1]
colors = colors[::-1]

plt.figure(figsize=(10,6))

bars = plt.barh(concerns, values, color=colors)

# Add values inside bars
for bar, value in zip(bars, values):
    plt.text(value - 0.3,
             bar.get_y() + bar.get_height()/2,
             f"{value}",
             va='center',
             ha='right',
             color='white',
             fontsize=12)

# Title and subtitle
plt.suptitle("Top 10 design concerns", fontsize=22, x=0.05, ha="left")
plt.text(0, len(concerns)+0.5, "concerns per 1,000", fontsize=14, color="gray")

# Remove x-axis
plt.xticks([])
plt.xlabel("")
plt.ylabel("")

# Clean style
for spine in plt.gca().spines.values():
    spine.set_visible(False)

plt.tight_layout()
plt.show()
```
