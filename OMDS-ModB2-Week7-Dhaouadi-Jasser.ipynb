```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size = numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Nitrate level (% of eutrophic value)")
```

```python
np.random.seed(1)
time_series = 100 * (1 - (np.random.normal(size = numdates) / 8 + np.arange(numdates) / 16))
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Phosphate level (% of eutrophic value)")
```

## Dataset 1: CDC Diabetes Health Indicators  (sourced from BRFSS) 

#### [Dataset 1] | *1- Load Preprocessed Dataset*

```python
dataset_path = "data/dataset1_diabetes_preprocessed_week2.csv"
df_diabetes = pd.read_csv(dataset_path)
```

#### [Dataset 1] | *2- General Health vs Diabetes Status*

```python
# Dataset 1: General Health vs Diabetes Status

plt.figure(figsize=(6,4))

sns.boxplot(
    data=df_diabetes,
    x="Diabetes_012",
    y="GenHlth"
)

plt.xticks([0,1,2], ["No Diabetes","Prediabetes","Diabetes"])
plt.xlabel("Diabetes Status")
plt.ylabel("General Health Score")
plt.title("General Health Status Across Diabetes Categories")

plt.tight_layout()
plt.show()
```

**Interpretation**

- Individuals with diabetes tend to report **poorer general health** compared to those without diabetes.

- The median health score increases (worse health) from **no diabetes to prediabetes to diabetes.**

- This suggests that diabetes is associated with **broader declines in perceived health and well-being**, not just metabolic indicators.

## Dataset 2: Estimation of Obesity Levels Based on Eating Habits and Physical Condition

#### [Dataset 2] | *1- Load Preprocessed Dataset*

```python
df_obesity = pd.read_csv("data/dataset2_obesity_preprocessed_week2_1.csv")
```

#### [Dataset 2] | *2- Vegetable Consumption vs Obesity Category*

```python
# Dataset 2: Vegetable Consumption vs Obesity Level

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df_obesity,
    x="NObeyesdad",
    y="FCVC"
)

plt.xticks(rotation=45)
plt.xlabel("Obesity Category")
plt.ylabel("Frequency of Vegetable Consumption (FCVC)")
plt.title("Vegetable Consumption Across Obesity Levels")

plt.tight_layout()
plt.show()
```

**Interpretation**

- Individuals in **lower obesity categories tend to report slightly higher vegetable consumption.**

- Higher obesity categories show **more variability and slightly lower median values**.

- While the relationship is not extremely strong, the pattern suggests **diet quality may influence weight outcomes**.

## Dataset 3: Cervical Cancer (Risk Factors)

#### [Dataset 3] | *1- Load Preprocessed Dataset*

```python
df_cervical_clean = pd.read_csv("data/dataset3_cervical_clean_preprocessed_week2.csv")
```

#### [Dataset 3] | *2- Smoking Years vs Age*

```python
# Dataset 3: Smoking Years vs Age

plt.figure(figsize=(6,4))

sns.scatterplot(
    data=df_cervical_clean,
    x="Age",
    y="Smokes (years)",
    alpha=0.5
)

plt.xlabel("Age")
plt.ylabel("Years of Smoking")
plt.title("Smoking Duration vs Age")

plt.tight_layout()
plt.show()
```

**Interpretation**

- Smoking duration increases with age, which is expected since older individuals have had more time to **accumulate smoking years**.

- Several individuals show **very long smoking histories**, which may represent long-term exposure to tobacco risk factors.

- Smoking is a known contributor to **immune suppression and cancer risk**, making this variable relevant when studying cervical cancer outcomes.
