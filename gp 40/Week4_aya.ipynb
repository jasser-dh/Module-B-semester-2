Below are a histogram and boxplot representation of the same data. A pharmacy is keeping a record of the prices of the drugs that it sells, and an administrator wants to know how much the more expensive drugs tend to cost, in the context of the other prices.

Please write a short explanation of the pros and cons of these two representations. Which would you choose? How would you modify the formatting, if at all, to make it more visually interesting, clear, or informative?

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


np.random.seed(0)
num_data = 100
data = np.exp(np.random.uniform(size = num_data) * 4)
df = pd.DataFrame(data.T, columns = ["data"])
```

```python
print("The 75th percentile is:", df.quantile(q = 0.75))
df.plot.hist()
```

```python
df.plot.box()
```

For this use case, let's compare the historgram vs the boxplot:
The histogram shows the "big picture" and the density of lower-priced drugs, but it can make the most expensive items look like a flat, insignificant line. The boxplot, conversely, is designed to highlight those expensive items by marking them as distinct outliers, though it sacrifices the visual "flow" of the price distribution. Since, the administrator is looking for the most expensive, then boxplot is definitely better (it shows outliers easily).

We can dfinitely enhance the output:

1) Add better labels and annotations for the data to make it informative of where the most expensive points are.
2) Use gradient colors to show the color scale of prices that would highlight the most expensive.
3) Overlay with jitter points to show the exact number of outliers (expensive items).

Here's a python code for these changes in the cell below.

```python

# 1. Re-create the user's data
np.random.seed(0)
num_data = 100
data = np.exp(np.random.uniform(size=num_data) * 4)
df = pd.DataFrame(data, columns=["Price"])

# 2. Set the visual theme for a professional look
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 7))

# 3. Create a Boxplot
# Using showmeans=True helps show the skewness relative to the median
sns.boxplot(
    x=df["Price"], 
    color="skyblue", 
    width=0.4, 
    showmeans=True,
    meanprops={"marker":"D", "markerfacecolor":"white", "markeredgecolor":"black", "markersize":"10"}
)

# 4. Overlay "Jittered" Data Points (Stripplot)
# This allows the administrator to see every single expensive drug individually
sns.stripplot(
    x=df["Price"], 
    color="#E67E22", # Professional orange
    alpha=0.6, 
    jitter=0.2, 
    size=6
)

# 5. Add informative formatting and annotations
plt.title("Price Distribution: Identifying High-Cost Drugs in Context", fontsize=18, fontweight='bold', pad=25)
plt.xlabel("Price ($)", fontsize=14, labelpad=15)
plt.yticks([]) # Hide Y-axis as it's not needed for a single distribution

# Calculate key statistics for annotations
median_val = df["Price"].median()
mean_val = df["Price"].mean()
max_val = df["Price"].max()

# Annotate Median
plt.annotate(f'Median: ${median_val:.2f}', xy=(median_val, -0.2), xytext=(median_val, -0.4),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=5),
             horizontalalignment='center', fontsize=11)

# Annotate Mean (to show the skew)
plt.annotate(f'Average (Mean): ${mean_val:.2f}', xy=(mean_val, 0.05), xytext=(mean_val + 5, 0.35),
             arrowprops=dict(facecolor='#2980B9', shrink=0.05, width=1, headwidth=5),
             horizontalalignment='left', fontsize=11, color='#2980B9', fontweight='bold')

# Annotate the most expensive drug (The Outlier)
plt.annotate(f'Max Price: ${max_val:.2f}', xy=(max_val, 0), xytext=(max_val - 2, -0.3),
             arrowprops=dict(facecolor='#C0392B', shrink=0.05, width=1, headwidth=5),
             horizontalalignment='right', color='#C0392B', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.savefig("drug_price_distribution.png")
```

I chose medical_insurance.csv to figure (Swarm, kde , violin) plots based on the python code in the next cell.

```python
import matplotlib.pyplot as plt
import seaborn as sns

print("Generating distribution plots for medical_insurance_df...")
medical_insurance_df = pd.read_csv('medical_insurance.csv')
for column in medical_insurance_df.columns:
    plt.figure(figsize=(10, 6)) # Create a new figure for each plot
    
    if pd.api.types.is_numeric_dtype(medical_insurance_df[column]):
        # Numerical column: histogram with KDE
        sns.histplot(medical_insurance_df[column].dropna(), kde=True)
        plt.title(f'Distribution of {column}')
        plt.xlabel(column)
        plt.ylabel('Frequency')
    else:
        # Categorical column: count plot
        sns.countplot(y=medical_insurance_df[column].dropna(), order=medical_insurance_df[column].value_counts().index, palette='viridis')
        plt.title(f'Count of {column}')
        plt.xlabel('Count')
        plt.ylabel(column)
        plt.xticks(rotation=45, ha='right') # Rotate x-axis labels if they are overlapping

    plt.tight_layout()
    plt.show()

print("Distribution plots for medical_insurance_df generated.")
```

In order to plot grouped histograms, we need to understand the patterns of the columns or features.
1st we dentify the numerical columns ('age', 'bmi', 'annual_medical_cost') and a categorical column ('sex') for grouping from medical_insurance_df to create grouped histograms.Then, we start plotting them together as followed in python code:

```python
import matplotlib.pyplot as plt
import seaborn as sns

print("Generating grouped histograms for 'age', 'bmi', and 'annual_medical_cost' by 'sex' in medical_insurance_df...")

columns_to_plot = ['age', 'bmi', 'annual_medical_cost']

if not medical_insurance_df.empty and 'sex' in medical_insurance_df.columns:
    for column in columns_to_plot:
        if column in medical_insurance_df.columns:
            plt.figure(figsize=(10, 6))
            sns.histplot(data=medical_insurance_df.dropna(subset=[column, 'sex']), x=column, hue='sex', kde=True, palette='viridis')
            plt.title(f'Distribution of {column.replace("_", " ").title()} by Sex')
            plt.xlabel(column.replace("_", " ").title())
            plt.ylabel('Frequency')
            plt.tight_layout()
            plt.show()
        else:
            print(f"Warning: Column '{column}' not found in medical_insurance_df. Skipping plot.")
else:
    print("medical_insurance_df is empty or 'sex' column is missing. Cannot generate grouped histograms.")

print("Grouped histograms generation complete.")
```

Analysis of grouping data:
The grouped histograms revealed distinct differences in the distributions of age, BMI, and annual medical cost between males and females.

**Age Distribution:** While both genders show a wide age range, their distributions might have subtle differences, potentially with varied peaks or spreads, though no specific numerical differences were highlighted in the provided output.
**BMI Distribution:** The distributions of BMI for males and females were observed to be distinct, suggesting potential differences in typical body mass index ranges or prevalence of certain BMI categories between the sexes.
**Annual Medical Cost Distribution:** The annual medical cost distributions also varied between males and females, indicating potential differences in healthcare expenditure patterns, average costs, or the prevalence of high-cost individuals across genders.

Bar plots are perfect for visualizing categorical columns, therefore we will run the following on medical_insurance.csv

```python
print("Generating bar plots for categorical variables in medical_insurance_df...")

categorical_cols_medical_insurance = medical_insurance_df.select_dtypes(include='object').columns

if not medical_insurance_df.empty and not categorical_cols_medical_insurance.empty:
    for column in categorical_cols_medical_insurance:
        plt.figure(figsize=(10, 6))
        # Use countplot for categorical data
        sns.countplot(y=medical_insurance_df[column].dropna(), order=medical_insurance_df[column].value_counts().index, palette='viridis', hue=medical_insurance_df[column].dropna(), legend=False)
        plt.title(f'Total Counts of {column}')
        plt.xlabel('Count')
        plt.ylabel(column)
        plt.tight_layout()
        plt.show()
else:
    print("medical_insurance_df is empty or contains no categorical columns. Cannot generate bar plots.")

print("Bar plots generation for medical_insurance_df complete.")
```

**Bin sizes critically affect how patterns are revealed.**
Small bin widths (many bins) can show fine details, minor peaks, and gaps in the data, but they can also make the histogram look noisy and emphasize random fluctuations, potentially making it harder to discern overall trends.
Large bin widths (few bins) smooth out the noise and show the broad shape of the distribution more clearly. However, they can obscure important details like modality (multiple peaks) or the presence of small clusters within the data.
Changing the bin width can definitely reveal different patterns. For instance, a very wide bin might make a bimodal distribution look unimodal, while narrower bins would separate the two peaks. Conversely, too narrow bins might make a perfectly smooth distribution appear jagged. Therefore, experimenting with bin sizes is an important step in exploratory data analysis.

# 4. Storytelling With Data graph

Reproduce any graph of your choice in p. 52-68 of the Storytelling With Data book as best you can.  (The second half of chapter two).  You do not have to get the exact data values right, just the overall look and feel.

One of the most impactful examples in that section (p. 52–68) is the transition from a cluttered, standard Excel bar chart to a clean, focused horizontal bar chart that highlights a specific data point.

Below is a reproduction of the "Concerns about the Environment" or "Reasons for Leaving" style charts found in those pages.

The Design Philosophy:
**Removed Chart Junk:** No borders, no gridlines, and no redundant Y-axis labels.

**Strategic Color:** Everything is muted (grey) except for the "story" we want to tell (blue).

**Clean Alignment:** Labels are placed directly next to the bars for easy reading.

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Create dummy data reflecting the "look and feel" of the book examples
categories = ['Price', 'Quality', 'Service', 'Location', 'Selection']
values = [45, 38, 30, 25, 12]
df = pd.DataFrame({'Category': categories, 'Value': values})
df = df.sort_values('Value', ascending=True) # Longest bars at the top for horizontal

# 2. Set up the figure
fig, ax = plt.subplots(figsize=(10, 6))

# 3. Plot - Using grey for "context" and blue for "focus"
colors = ['#D3D3D3'] * len(df)
colors[-1] = '#1F77B4'  # Highlight the top category (Price) in bold blue

bars = ax.barh(df['Category'], df['Value'], color=colors, height=0.7)

# 4. Decluttering (The "Storytelling with Data" way)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['bottom'].set_visible(False)
ax.spines['left'].set_visible(False)

# Remove x-axis ticks/labels (we will label bars directly)
ax.get_xaxis().set_visible(False)

# Adjust y-axis font
ax.tick_params(axis='y', which='major', labelsize=12, color='white')

# 5. Add direct labeling
for i, bar in enumerate(bars):
    width = bar.get_width()
    label_color = '#1F77B4' if i == len(bars)-1 else '#555555'
    font_weight = 'bold' if i == len(bars)-1 else 'normal'
    
    ax.text(width + 1, bar.get_y() + bar.get_height()/2, 
            f'{int(width)}%', 
            va='center', fontsize=12, color=label_color, fontweight=font_weight)

# 6. Title and Source (Professional formatting)
plt.title('Price remains the primary concern for our customers', 
          fontsize=16, loc='left', pad=20, color='#333333', fontweight='bold')

plt.tight_layout()
plt.show()
```

Conclusion:

**Horizontal Orientation:** By switching from vertical to horizontal, the category labels are much easier to read without tilting your head.

**Enclosure/Color:** By making the "Price" bar blue and the others grey, you are telling the administrator exactly where to look before they even process the numbers.

**Data-to-Ink Ratio:** By removing the gridlines and the X-axis scale, we maximize the focus on the data itself.
